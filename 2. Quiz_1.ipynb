### 10. **Conceptual:** Do I need to clear the gradients before or after calling backward? Why?
<details><summary>Answer</summary>
You must clear gradients **before** calling `loss.backward()`, because PyTorch accumulates gradients by default — so failing to reset them means you’d be summing gradients from multiple passes.
</details>
