## 🔹 Neural Network Basics

**1. A neural network is called *deep* when it has:**
a) 1 layer  
b) 2 or more layers  
c) 5 or more layers  
d) Hidden layers only  

<details><summary>✅ Answer</summary>
b) 2 or more layers
</details>

---

**2. The main reason for using activation functions is to:**
a) Reduce training time  
b) Add non-linearity  
c) Prevent overfitting  
d) Normalize data  

<details><summary>✅ Answer</summary>
b) Add non-linearity
</details>

---

**3. Which activation is typically used in the output layer for multi-class classification?**  
a) Sigmoid  
b) Tanh  
c) Softmax  
d) ReLU

</details>

## 🔹 Overfitting vs Underfitting

</details>

---
